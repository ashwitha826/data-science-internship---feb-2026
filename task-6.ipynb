```python
#Smart Parking Lot Management System
def parking_system(capacity, logs):
    current = 0
    peak = 0
    for log in logs:
        if log == "IN":
            current += 1
        elif log == "OUT" and current > 0:
            current -= 1
        if current > peak:
            peak = current
    status = "Full" if current >= capacity else "Available"
    print("Currently Parked Vehicles:", current)
    print("Parking Status:", status)
parking_system(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
#Online Food Delivery Time Estimator
def estimate_time(distance,traffic,weather):
    time=distance*5
    if traffic.lower()=="high":
        time+=15
    elif traffic.lower()=="medium":
        time+=10
    if weather.lower()=="rainy":
        time+=10
    elif weather.lower()=="stormy":
        time+=20
    return time
distance=float(input())
traffic=input()
weather=input()
eta=estimate_time(distance, traffic, weather)
print("Estimated Delivery Time:", int(eta), "minutes")
```

```python
#Movie Theatre Seat Occupancy Analyzer
def costcal(seats,occupancy):
    total=0  # count occupied seats
    for occupied in occupancy:
        if(occupied):
            total+=1
    per=(total*100)/seats  # calculate occupancy percentage
    print("Occupancy percentage is: ",per)
    if(per==100):
        print("HouseFull ,open additional shows")
    elif(per>=75 and per<=99):
        print("Almost Full")
    else:
        print("Seats available")
seats=int(input("Enter the total no of seats: "))
occupancy=list(map(int,input("Enter the occupancy 1 for occupied and 0 for not occupied separated by a space").split()))
costcal(seats,occupancy)
```

```python
#Cloud Server Load Classification System
def server_load(cpu_readings):
    total = 0
    for reading in cpu_readings:
        total += reading
    avg = total / len(cpu_readings)
    if avg < 50:
        status = "Normal"
    elif avg <= 80:
        status = "Warning"
    else:
        status = "Critical"
    print("Average CPU Load:", round(avg), "%")
    print("Server Status:", status)
server_load([45, 60, 70, 85, 90])
```

```python
#Smart Classroom Resource Usage Monitor
def findoveruse(Resource_Usage):
    item=None  # store overused resource
    val=0  # store highest usage
    for resource in Resource_Usage:
        if(val<Resource_Usage[resource]):
            val=Resource_Usage[resource]
            item=resource
    print("Overused Resource is: ",item)
    if(val>6 and val<=12): #assuming classroom is of 12hrs
        print("Energy Alert: Yes")
    else:
        print("Energy Alert: No")
Resource_Usage={
 "Projector": 6,
 "AC": 9,
 "Lights": 4
}
findoveruse(Resource_Usage)
```

```python
#Online Event Registration Capacity Controller
def registration_system(capacity, registrations):
    confirmed = min(capacity, registrations)
    waitlisted = max(0, registrations - capacity)
    status = "Closed" if registrations >= capacity else "Open"
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)

registration_system(100, 105)
```
